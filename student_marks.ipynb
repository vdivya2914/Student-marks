```python
marks = [45, 55, 65, 75, 85, 95]

key = int(input("Enter marks to search: "))

low = 0
high = len(marks) - 1

while low <= high:
    mid = (low + high) // 2

    if marks[mid] == key:
        print("Marks found")
        break
    elif marks[mid] < key:
        low = mid + 1
    else:
        high = mid - 1
else:
    print("Marks not found")
```
